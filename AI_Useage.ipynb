```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('AI_Usage_and_Impact_on_Students_and_Professionals.csv')
data.head()
```

```python
data['Age'] = data['Age'].str.replace('yrs' , '')
```

```python
from sklearn.impute import SimpleImputer
```

```python
imputers = SimpleImputer(strategy ='median')
imputers.fit(data[['Age']])
data[['Age']] = imputers.transform(data[['Age']])
```

```python
data['Gender'].unique()
```

```python
data['Gender'] = data['Gender'].str.strip()
data['Gender'] = data['Gender'].str.title()
```

```python
data['Gender'] = data['Gender'].replace({
    'M': 'Male',
    'F': 'Female'
})
```

```python
data['Gender'].unique()
```

```python
imputers = SimpleImputer(strategy ='most_frequent')
imputers.fit(data[['Gender']])
data[['Gender']] = imputers.transform(data[['Gender']])
```

```python
imputers = SimpleImputer(strategy ='most_frequent')
imputers.fit(data[['Country']])
data[['Country']] = imputers.transform(data[['Country']])
```

```python
data['Education_Level'].unique()
```

```python
imputers = SimpleImputer(strategy ='most_frequent')
imputers.fit(data[['Education_Level']])
data[['Education_Level ']] = imputers.transform(data[['Education_Level']])
```

```python
data['Profession'] = data['Profession'].str.strip()
```

```python
data['Profession'] = data['Profession'].str.title()
```

```python
data['Monthly_Income'] = pd.to_numeric(data['Monthly_Income'], errors='coerce')


imputers = SimpleImputer(strategy = 'median')
imputers.fit(data[['Monthly_Income']])
data[['Monthly_Income']] = imputers.transform(data[['Monthly_Income']])
```

```python
data['Monthly_Income'].isnull().sum()
```

```python
data['AI_Tool'] = data['AI_Tool'].str.strip()
data['AI_Tool'] = data['AI_Tool'].str.title()
```

```python
imputers = SimpleImputer(strategy = 'most_frequent')
imputers.fit(data[['AI_Tool']])
data[['AI_Tool']] = imputers.transform(data[['AI_Tool']])
```

```python
ai_tool_mapping = {
    'Chatgpt': 'ChatGPT',
    'Chat Gpt': 'ChatGPT',
    'Gemini': 'Google Gemini',
    'Copilot': 'Microsoft Copilot'
}

data['AI_Tool'] = data['AI_Tool'].replace(ai_tool_mapping)
```

```python
data['AI_Tool'].value_counts()
```

```python
data['AI_Tool'].isnull().sum()
```

```python
data['AI_Purpose'] = data['AI_Purpose'].fillna('Unknown')
```

```python
data['AI_Usage_Hours_Per_Day'].str.replace('hrs','')
data['AI_Usage_Hours_Per_Day'] = pd.to_numeric(
    data['AI_Usage_Hours_Per_Day'], errors='coerce'
)
```

```python
data['AI_Usage_Hours_Per_Day'] = data['AI_Usage_Hours_Per_Day'].fillna(
    data['AI_Usage_Hours_Per_Day'].median()
)
data['AI_Usage_Hours_Per_Day'].isnull().sum()
```

```python
data['Would_Recommend'] = data['Would_Recommend'].fillna(
    data['Would_Recommend'].mode()[0]
)
data['Would_Recommend'].isnull().sum()
```

```python
cols = ['Time_Saved_Hours_Per_Week', 'Work_or_Study_Hours_Per_Day']

for col in cols:
    data[cols] = data[cols].str.replace('hrs', '', regex=False)
```

```python
cols = ['Time_Saved_Hours_Per_Week', 'Work_or_Study_Hours_Per_Day']

data[cols] = data[cols].apply(pd.to_numeric, errors='coerce')
```

```python
data['Time_Saved_Hours_Per_Week'] = data['Time_Saved_Hours_Per_Week'].fillna(
    data['Time_Saved_Hours_Per_Week'].median()
)

data['Work_or_Study_Hours_Per_Day'] = data['Work_or_Study_Hours_Per_Day'].fillna(
    data['Work_or_Study_Hours_Per_Day'].mean()
)
```

```python
# Categorical columns
cat_cols = ['Education_Level', 'Tasks_Performed']

data[cat_cols] = data[cat_cols].fillna('Unknown')


# Numerical columns
num_cols = [
    'Monthly_AI_Cost',
    'Productivity_Score',
    'Accuracy_Rating',
    'Satisfaction_Score'
]


for col in num_cols:
    data[col] = pd.to_numeric(data[col], errors='coerce')
```
